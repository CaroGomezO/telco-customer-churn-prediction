# INTRODUCCION
-----
## Problema seleccionado:
**Churn o pérdida de clientes en empresas de telecomunicaciones**:

El churn o pérdida de clientes es un problema crítico para la industria de telecomunicaciones. Cada cliente que se va representa ingresos recurrentes perdidos y un costo de reemplazo que suele ser mayor que el de simplemente retenerlo. En mercados con millones de usuarios reducir la tasa de abandono incluso en un 1% puede traducirse en ganancias significativas.

Este problema es especialmente pertinente en nuestro contexto, porque existe una fuerte competencia entre operadores de telecomunicaciones (Claro, Movistar, Tigo, WOM, entre otros) y donde los usuarios están constantemente expuestos a spam de llamadas y mensajes de ofertas de la competencia buscando captar clientes insatisfechos.


## Objetivo del modelo:
**Predecir el comportamiento para retener a los clientes.**:

El objetivo es construir un modelo que prediga si un cliente cancelará el servicio, de modo que, a partir de sus características, la empresa pueda dirigir acciones de retención antes de que el cliente se vaya. Esto sería bastante útil en nuestro país, ya que permitiría a los operadores anticiparse al abandono, priorizar clientes en riesgo y diseñar ofertas personalizadas, programas de fidelización o mejoras en el servicio que reduzcan la pérdida de usuarios y protejan sus ingresos recurrentes.


## Tipo de problema:
**Clasificación binaria**
Aprendizaje supervisado de clasificación binaria. Se busca predecir si un cliente abandona el servicio (Churn = Sí/1) o no (Churn = No/0). No es un problema de regresión porque la salida es categórica. Se puede usar un modelo como regresión logística para estimar la probabilidad de churn y, mediante un umbral, convertir esa probabilidad en una clase.

## variable objetivo:
**Churn**
Churn (Yes/No o 1/0). Es una variable categórica binaria que indica si el cliente canceló el servicio o no.

## Fuente de los datos:

El dataset utilizado es Telco Customer Churn, publicado originalmente por IBM como parte de sus Sample Data Sets. Contiene información de 7043 clientes de una empresa ficticia de telecomunicaciones en California durante el tercer trimestre (Q3). Incluye variables demográficas, servicios contratados, tipo de contrato, forma de pago, facturación y la variable objetivo Churn.
Enlace del dataset:
[Telco Customer Churn: IBM / Kaggle](https://www.kaggle.com/datasets/blastchar/telco-customer-churn/data)


# Descripción del conjunto de datos
----
#### Número de observaciones: 7043 
#### Número de variables: 21
**customerID** (Identificador): Identificador único de cada cliente y no sirve como variable predictora.
**gender** (Categórica nominal): 	Género del cliente: masculino o femenino.
**SeniorCitizen** (Categórica binaria): Indica si el cliente es adulto mayor: 1 = sí, 0 = no.
**Partner** (Categórica binaria):  Indica si el cliente tiene pareja: Yes / No.
**Dependents** (Categórica binaria): Indica si el cliente tiene dependientes: Yes / No.------------
**tenure** (Numérica discreta): Número de meses que el cliente lleva con la compañía.
**PhoneService** (Categórica binaria): Indica si el cliente tiene servicio telefónico: Yes / No.
**MultipleLines** (Categórica nominal): Indica si el cliente tiene múltiples líneas: Yes / No / No phone service.
**InternetService** (Categórica nominal): Proveedor de internet del cliente: DSL / Fiber optic / No.
**OnlineSecurity** (Categórica nominal): Indica si tiene seguridad en línea: Yes / No / No internet service.
**OnlineBackup** (Categórica nominal): Indica si tiene respaldo en línea: Yes / No / No internet service.
**DeviceProtection** (Categórica nominal): Indica si tiene protección de dispositivos: Yes / No / No internet service.
**TechSupport** (Categórica nominal): Indica si tiene soporte técnico: Yes / No / No internet service.
**StreamingTV** (Categórica nominal): Indica si tiene TV en streaming: Yes / No / No internet service.
**StreamingMovies** (Categórica nominal): Indica si tiene películas en streaming: Yes / No / No internet service.
**Contract** (Categórica nominal): Tipo de contrato del cliente: Month-to-month / One year / Two year.
**PaperlessBilling** (Categórica binaria): Indica si tiene facturación electrónica: Yes / No.
**PaymentMethod** (Categórica nominal): Método de pago: Electronic check / Mailed check / Bank transfer (automatic) / Credit card (automatic).
**MonthlyCharges** (Numérica continua): Monto que se le cobra al cliente cada mes.
**TotalCharges** (Numérica continua): Monto total acumulado cobrado al cliente. En pandas aparece como str porque contiene valores vacíos.
**Churn** (Categórica binaria).

#### Variable objetivo: 
**Churn**: Es una variable categórica binaria que indica si el cliente canceló el servicio o no.

#### Posibles limitaciones del conjunto de datos:

Variables categóricas: Una limitación importante es que la mayoría de las variables son categóricas, por lo que deben ser tratadas de manera diferente a las variables numéricas antes de entrenar modelos que no manejan texto.

Escalas diferentes: variables numéricas como MonthlyCharges, TotalCharges y tenure pueden requerir normalización o estandarización según el modelo que vayamos a utilizar.

customerID no aporta: es un identificador único y debe eliminarse antes de entrenar el modelo.
